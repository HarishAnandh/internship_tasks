```python
import pandas as pd

# Read CSV file
data = pd.read_csv("trainingdata.csv")

concepts = data.iloc[:, :-1].values
target = data.iloc[:, -1].values

def candidate_elimination(concepts, target):

    specific = concepts[0].copy()

    general = [['?' for _ in range(len(specific))]
               for _ in range(len(specific))]

    for i, h in enumerate(concepts):
    
        if target[i] == "Yes":
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    specific[x] = '?'
                    general[x][x] = '?'

        elif target[i] == "No":
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    general[x][x] = specific[x]
                else:
                    general[x][x] = '?'

    general = [g for g in general if g != ['?'] * len(specific)]

    return specific, general

s, g = candidate_elimination(concepts, target)

print("Specific Hypothesis:")
print(s)

print("\nGeneral Hypotheses:")
for hypothesis in g:
    print(hypothesis)
```
